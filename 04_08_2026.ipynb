```python
#Exp.5.1 - Discretization and Binning
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/salemprakash/EDA/main/Data/sales.csv')

print("DataFrame columns:", df.columns)
print()
df['Sales'] = df['UnitPrice'] * df['Quantity']

print(df['Sales'].describe())
print()

bins = [-1, 10, 100, 500, 1000, 5000, 30000]
bin_labels = ['Very Low', 'Low', 'Medium', 'High', 'Very High', 'Extreme']

df['Amount_Bin'] = pd.cut(df['Sales'], bins=bins, labels=bin_labels)

print(pd.Series(df['Amount_Bin']).value_counts())
print()
df['Amount_Bin_Equal'] = pd.cut(df['Sales'], bins=5, precision=2)
print(pd.Series(df['Amount_Bin_Equal']).value_counts())
print()
df['Amount_Bin_Quantile'] = pd.qcut(df['Sales'], q=4, duplicates='drop')
print(pd.Series(df['Amount_Bin_Quantile']).value_counts())
print()
bin_counts = df['Amount_Bin'].value_counts()
outlier_bins = bin_counts[bin_counts < bin_counts.mean() * 0.05]
print("Outlier bins:\n", outlier_bins)
print()
print(df.groupby('Amount_Bin', observed=True)['Sales'].mean())
print()
```

```python
#Exp.5.1 - Discretization and Binning
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler

df_num = df[['Sales']]

minmax = MinMaxScaler()
df_minmax = pd.DataFrame(
minmax.fit_transform(df_num),
columns=['Sales_MinMax']
)
print("Min-Max Scaled:\n", df_minmax.head())
print()

standard = StandardScaler()
df_std = pd.DataFrame(
standard.fit_transform(df_num),
columns=['Sales_Std']
)
print("Standardized:\n", df_std.head())
print()

robust = RobustScaler()
df_robust = pd.DataFrame(
robust.fit_transform(df_num),
columns=['Sales_Robust']
)
print("Robust Scaled:\n", df_robust.head())
print()
df_final = pd.concat([df, df_minmax, df_std, df_robust], axis=1)
print(df_final.head())
```

```python
#Exp.5.2 - Feature Scaling and Normalization
import pandas as pd
import numpy as np

df = pd.read_csv('https://raw.githubusercontent.com/salemprakash/EDA/main/Data/sales.csv')

df['Sales'] = df['UnitPrice'] * df['Quantity']

print("Duplicate rows found:", df.duplicated().sum())
print()
df = df.drop_duplicates()
print("Shape after removing duplicates:", df.shape)
print()

def amount_category(x):
    if x < 50:
        return 'Small'
    elif x < 500:
        return 'Medium'
    else:
        return 'Large'


df['Sales_Category'] = df['Sales'].map(amount_category)
print(df['Sales_Category'].value_counts())
print()

df['Sales_Log'] = np.log1p(df['Sales'])

print("Original Sales stats:\n", df['Sales'].describe())
print()
print("Log-Transformed Sales stats:\n", df['Sales_Log'].describe())
print()
df['Sales_Log_Bin'] = pd.cut(df['Sales_Log'], bins=5)
print(pd.Series(df['Sales_Log_Bin']).value_counts())
print()
```

```python
#Exp.5.3 - Outlier Detection
import pandas as pd
import numpy as np
from scipy import stats


df = pd.read_csv('https://raw.githubusercontent.com/salemprakash/EDA/main/Data/sales.csv')


df['Sales'] = df['UnitPrice'] * df['Quantity']

Q1 = df['Sales'].quantile(0.25)
Q3 = df['Sales'].quantile(0.75)
IQR = Q3 - Q1

print(f"Q1 (Sales): {Q1}, Q3 (Sales): {Q3}, IQR (Sales): {IQR}")
print()
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"Lower Bound (Sales): {lower_bound}, Upper Bound (Sales): {upper_bound}")
print()
outliers_iqr = df[(df['Sales'] < lower_bound) | (df['Sales'] > upper_bound)]
print("Number of outliers (IQR method):", outliers_iqr.shape[0])
print()
print(outliers_iqr[['Sales']].head())
print()
df_clean_iqr = df[(df['Sales'] >= lower_bound) & (df['Sales'] <= upper_bound)]
print("Shape after removing outliers (IQR method):", df_clean_iqr.shape)
print()

df['Sales_Zscore'] = stats.zscore(df['Sales'])

outliers_z = df[np.abs(df['Sales_Zscore']) > 3]
print("Number of outliers (Z-score method):", outliers_z.shape[0])
print()
print(outliers_z[['Sales', 'Sales_Zscore']].head())
print()
df_clean_z = df[np.abs(df['Sales_Zscore']) <= 3]
print("Shape after removing z-score outliers (Z-score method):", df_clean_z.shape)
print()
```
