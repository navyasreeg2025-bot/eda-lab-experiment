```python
import pandas as pd
import numpy as np

data = np.arange(10, 25).reshape(5, 3)
df = pd.DataFrame(data,index=['Raj','Priya','Amit','Sara','Neha'],columns=['Math','Science','English'])
```

```python
df['History'] = np.nan
df.loc['Kiran'] = [10,11,12,13, np.nan]
df.loc['Vikas'] = np.nan
df['Geography'] = np.nan
df.loc['Raj','History'] = 15
print(df)
print(df.columns)
```

```python
df.dropna()
df.dropna(how='all')
df.dropna(how='all', axis=1)
df.dropna(thresh=5, axis=1)
```

```python
filledDf = df.fillna(0)
df.History.fillna(method='ffill')
df.History.fillna(method='bfill')
```
