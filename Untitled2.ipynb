```python
import pandas as pd
import numpy as np


df1_cricket = pd.DataFrame({
'PlayerID': [101, 103, 105, 107, 109],
'RunsScored': [45, 78, 12, 90, 33]
})
df2_cricket = pd.DataFrame({
'PlayerID': [102, 104, 106, 108, 110],
'RunsScored': [56, 21, 67, 88, 40]
})


df1_football = pd.DataFrame({
'PlayerID': [101, 103, 105, 107, 109],
'GoalsScored': [2, 0, 1, 3, 0]
})
df2_football = pd.DataFrame({
'PlayerID': [102, 104, 106],
'GoalsScored': [1, 4, 0]
            })
```

```python
df_cricket = pd.concat([df1_cricket, df2_cricket], ignore_index=True)

df_football = pd.concat([df1_football, df2_football], ignore_index=True)

print(df_cricket)
```

```python
df_inner = df_cricket.merge(df_football, on='PlayerID', how='inner')
df_left = df_cricket.merge(df_football, on='PlayerID', how='left')
df_right = df_cricket.merge(df_football, on='PlayerID', how='right')

print(df_inner)
print()
print(df_left)
print()
print(df_right)
```
