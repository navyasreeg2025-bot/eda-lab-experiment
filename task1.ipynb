```python
import pandas as pd


df = pd.read_csv("employee_salary.csv")

print("\n========== TASK 1 ==========\n")


print("First 5 Records:")
print(df.head())


print("\nLast 5 Records:")
print(df.tail())

print("\nDataset Shape:")
print(df.shape)

rows, columns = df.shape

print("\nNumber of Records:", rows)
print("Number of Features:", columns)



print("\n========== TASK 2 ==========\n")

print("Column Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nDataset Information:")
df.info()

print("\nDescriptive Statistics:")
print(df.describe())

print("\nAverage Employee Salary:")
print(df["Salary"].mean())

print("\nStandard Deviation:")
print(df.std(numeric_only=True))

std = df.std(numeric_only=True)

print("\nFeature with Highest Standard Deviation:")
print(std.idxmax(), "=", std.max())



print("\n========== TASK 3 ==========\n")

print("Missing Values in Each Column:")
print(df.isnull().sum())

print("\nMissing Value Percentage:")
print((df.isnull().sum()/len(df))*100)

print("\nColumn with Highest Missing Values:")
print(df.isnull().sum().idxmax())

duplicates = df.duplicated().sum()

print("\nNumber of Duplicate Records:")
print(duplicates)

print("\nDataset Size Before Cleaning:")
print(df.shape)

df_clean = df.drop_duplicates()

print("\nDataset Size After Cleaning:")
print(df_clean.shape)

df_clean.to_csv("employee_salary_cleaned.csv", index=False)

print("\nCleaned dataset saved as employee_salary_cleaned.csv")
```
